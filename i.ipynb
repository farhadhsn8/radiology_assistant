```python
from pydub import AudioSegment

def convert_to_mp3(input_file_path, output_file_path):
    """
    Convert an audio file to MP3 format.

    :param input_file_path: Path to the input audio file.
    :param output_file_path: Path where the output MP3 file will be saved.
    """
    # Load the audio file
    audio = AudioSegment.from_file(input_file_path)
    # Export as MP3
    audio.export(output_file_path, format="mp3")

convert_to_mp3("Default_20250718-125325.m4a", "out.mp3")
```

```python
import torch
from transformers import pipeline



device =  -1
        
pipe = pipeline(
    task="automatic-speech-recognition",
    model="openai/whisper-base",  # facebook/wav2vec2-large-robust    openai/whisper-base      facebook/wav2vec2-base-960h
    device=device,
)


def speech_to_text(
    audio_path: str,
    model_name: str = "openai/whisper-base",
    language: str = "en"
) -> str:
    """
    Convert speech from an audio file to text using Hugging Face Transformers.
    
    Args:
        audio_path: Path to the audio file (.wav, .mp3, .flac, etc.)
        model_name: Hugging Face model identifier (default: "openai/whisper-base")
        language: Language code (e.g., 'en' for English, 'fr' for French)
    
    Returns:
        Transcribed text from the audio
    
    Requires:
        pip install torch transformers
    """
    # Initialize pipeline on first call
    
    
    # Process audio with language specification
    result = pipe(
        audio_path,
        generate_kwargs={"language": language}
    )
    return result["text"]
```

```python
# Example usage
text = speech_to_text("تست ۶.mp3")
print("Transcribed text:", text)
```
