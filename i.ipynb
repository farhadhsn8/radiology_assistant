```python
import torch
from transformers import pipeline



device =  -1
        
pipe = pipeline(
    task="automatic-speech-recognition",
    model="openai/whisper-base",  # facebook/wav2vec2-large-robust    openai/whisper-base      facebook/wav2vec2-base-960h
    device=device,
)


def speech_to_text(
    audio_path: str,
    model_name: str = "openai/whisper-base",
    language: str = "en"
) -> str:
    """
    Convert speech from an audio file to text using Hugging Face Transformers.
    
    Args:
        audio_path: Path to the audio file (.wav, .mp3, .flac, etc.)
        model_name: Hugging Face model identifier (default: "openai/whisper-base")
        language: Language code (e.g., 'en' for English, 'fr' for French)
    
    Returns:
        Transcribed text from the audio
    
    Requires:
        pip install torch transformers
    """
    # Initialize pipeline on first call
    
    
    # Process audio with language specification
    result = pipe(
        audio_path,
        generate_kwargs={"language": language}
    )
    return result["text"]
```

```python
# Example usage
text = speech_to_text("Default_20250718-125325.mp3")
print("Transcribed text:", text)
```
